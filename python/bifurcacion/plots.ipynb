```python
import numpy as np 
import pandas as pd 
from scipy.signal import find_peaks
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
import jax.numpy as jnp
```

```python
xs = pd.read_pickle('xs2.pkl').to_numpy()
ys = pd.read_pickle('ys2.pkl').to_numpy()
#BETAS PARA LA TIRADA 1 XD 
# betas = np.arange(22.8, 50, 0.01)
# ts = np.linspace(1000, 1150, 20000)

#BETAS PARA LA TIRADA 2 XD
betas = jnp.arange(22.8,38, 0.005)
jnp.linspace(1000, 1150, 60000)
```

```python
peak_heights = {}

for n,b in enumerate(betas):
    peaks_vals = np.unique(find_peaks(xs[n,::], height=0)[1]['peak_heights'])
    peak_heights[b] = peaks_vals[peaks_vals>1]
```

```python
plt.figure(figsize=(14,4))

for b, heights in peak_heights.items():
    plt.plot([b]*len(heights), heights, '.', color='black', markersize=1)
    
# plt.xlim(34,22.7)
```
