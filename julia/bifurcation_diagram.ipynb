```julia
heights_per_r = Vector{Vector{Float32}}(undef, length(Rs));
```

```julia
for n in 1:length(Rs)
    peaks, _ = findpeaks1d(xs[n,:],relheight=0)
    peak_heights = xs[n,:][peaks]
    # if n > 20842
    #     peak_heights =[mean(xs[n,:])]
    # end
    heights_per_r[n] = peak_heights
end
```

```julia
f = Figure()
ax = Axis(f[1, 1])

for n in 1:length(Rs)
    peaksinn =heights_per_r[n]
    
    points = Point2f.(Rs[n],peaksinn)
    
    scatter!(points,color=:black,marker=:circle,markersize=3)
end
xlims!(ax,1940,1910)
# ylims!(ax,-4,1)
f
```
